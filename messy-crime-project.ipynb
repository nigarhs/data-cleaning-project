```python
import re

def clean_phone(val):
    val = str(val)
    if val.lower() in ['nan', 'unknown']:
        return 'Unknown'
    digits = re.sub(r'\D', '', val)  # strip everything except digits
    if len(digits) == 10:
        return f"{digits[0:3]}-{digits[3:6]}-{digits[6:10]}"
    return 'Unknown' 

df['victim_phone'] = df['victim_phone'].apply(clean_phone)
```

```python
# Start fresh from the original column
original = pd.read_csv("C:/Users/Vivobook/Downloads/python-project/crime_incidents_messy.csv")
df['incident_datetime'] = original['incident_datetime']

# Identify each pattern precisely
iso_mask = df['incident_datetime'].str.match(r'^\d{4}-\d{2}-\d{2}', na=False)      # 2024-04-16 08:45:03
dash_mask = df['incident_datetime'].str.match(r'^\d{2}-\d{2}-\d{4}', na=False)     # 05-03-2020 (DD-MM-YYYY)
slash_mask = df['incident_datetime'].str.contains('/', na=False)                   # 04/01/2020 03:09 (MM/DD/YYYY)

# Parse each group with the correct rule
df.loc[iso_mask, 'incident_datetime'] = pd.to_datetime(df.loc[iso_mask, 'incident_datetime'], errors='coerce')
df.loc[dash_mask, 'incident_datetime'] = pd.to_datetime(df.loc[dash_mask, 'incident_datetime'], dayfirst=True, errors='coerce')
df.loc[slash_mask, 'incident_datetime'] = pd.to_datetime(df.loc[slash_mask, 'incident_datetime'], dayfirst=False, errors='coerce')

# Final conversion to proper datetime dtype
df['incident_datetime'] = pd.to_datetime(df['incident_datetime'], errors='coerce')

print(df['incident_datetime'].isnull().sum())
print(df['incident_datetime'].head())
```

```python
# Start fresh from the original column
original = pd.read_csv("C:/Users/Vivobook/Downloads/python-project/crime_incidents_messy.csv")
raw = original['incident_datetime']

# Identify each pattern precisely
iso_mask = raw.str.match(r'^\d{4}-\d{2}-\d{2}', na=False)
dash_mask = raw.str.match(r'^\d{2}-\d{2}-\d{4}', na=False)
slash_mask = raw.str.contains('/', na=False)

# Parse each group into a NEW datetime series (don't touch df yet)
parsed = pd.Series(pd.NaT, index=raw.index)
parsed[iso_mask] = pd.to_datetime(raw[iso_mask], errors='coerce')
parsed[dash_mask] = pd.to_datetime(raw[dash_mask], dayfirst=True, errors='coerce')
parsed[slash_mask] = pd.to_datetime(raw[slash_mask], dayfirst=False, errors='coerce')

# Now assign the fully-built datetime series to df in one go
df['incident_datetime'] = parsed

print(df['incident_datetime'].isnull().sum())
print(df['incident_datetime'].head())
```
